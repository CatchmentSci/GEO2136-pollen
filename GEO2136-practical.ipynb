## 7. Add a legend and refine the figure
The function below groups the drawing commands together. Calling `draw_pollen()` displays the figure. We will also use it to export the same figure.

**Tasks:**
1. Match the legend colours to the colours you actually assigned.
2. Add `"Charcoal"` to `legend` and its colour to `fill`.
3. Adjust `inset` to position the legend; check that it and all labels are visible.
4. Choose readable bar widths, label angles and colours.

Whenever you edit this function, rerun the cell to update its definition.

See the [riojaPlot documentation and galleries](https://nsj3.github.io/riojaPlot/) for further examples. Aim for a clear figure suitable for your report, with your own presentation choices.

```r
draw_pollen <- function() {
  old_par <- par(no.readonly = TRUE)
  on.exit(par(old_par))
  par(oma = c(0, 0, 3, 0))

  riojaPlot(
    pollen_data, sample_names,
    plot.poly = FALSE, plot.line = FALSE,
    plot.bar = TRUE, lwd.bar = 10,
    scale.percent = TRUE, srt.xlabel = 45,
    col.bar = p.col.group
  )

  # Match these colours to your assignments and add Charcoal.
  legend(
    "top", inset = c(0, -0.19), xpd = NA,
    legend = c("Trees", "Shrubs", "Herbs", "Wetland", "Ferns and mosses"),
    fill = c("darkgrey", "blue", "orange", "red", "purple"),
    horiz = TRUE, bty = "n", cex = 0.8
  )
}

draw_pollen()
```

## 8. Export the pollen diagram
This cell writes a high-resolution PNG. Find it in Jupyter's file browser, open it to check the layout, then right-click and choose **Download**.

Rerun this cell after making changes to update the exported file. `dev.off()` closes the image device so the file is written properly.

```r
png("pollen_diagram.png", width = 16, height = 8, units = "in", res = 300)
tryCatch(draw_pollen(), finally = dev.off())
```
